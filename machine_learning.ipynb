```python
# matrix (11-50),elements -----> martrix 4*4-----> rw wise sum ,column wise ,total sum
# matrix (11-20), matrix(31-40)----->reshape(2*5)-----> a+b.
```

```python
# matrix (11-50),elements -----> martrix 4*4-----> rw wise sum ,column wise ,total sum

a = np.random.randint(11, 50, 16).reshape(4, 4)

#print("Matrix:,a")
print(a)

print("Row-wise sum:")
print(np.sum(a, axis=1))

print("Column-wise sum:")
print(np.sum(a, axis=0))

print("Total sum:")
print(np.sum(a))
```

```python
# matrix (11-20), matrix(31-40)----->reshape(2*5)-----> a+b.
import numpy as np

a = np.arange(11, 21).reshape(2, 5)
b = np.arange(31, 41).reshape(2, 5)

print("Matrix A:")
print(a)

print("Matrix B:")
print(b)

print("A + B:")
print(a + b)
```
